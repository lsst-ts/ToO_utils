```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

allMDCData = pd.read_csv("../mdc_data/mdc_Rubin_EFD_data.csv")
```

# Make a plot of received alerts by time during the MDC

```python
timeArr = np.arange(np.datetime64("2025-09-25"), np.datetime64("2025-11-20"))
receivedArr = [np.datetime64(x) for x in allMDCData["EFD datetime"]]
cumulative = []

for t in timeArr:
    msk = receivedArr < t
    cumulative.append(len(np.where(msk)[0]))
```

```python
fig, ax = plt.subplots(dpi=300)
ax.plot(timeArr, cumulative, "--", color="blue")
ax.fill_between(
    timeArr, cumulative, where=np.array(cumulative) >= 0, facecolor="blue", alpha=0.5
)
ax.grid(ls="-")
ax.set_xticks(
    np.arange(
        np.datetime64("2025-09-25"),
        np.datetime64("2025-11-20"),
        step=np.timedelta64(10, "D"),
    )
)
ax.xaxis.set_major_formatter(mdates.DateFormatter("%b %d"))
ax.set_ylim(0, 225)
ax.set_xlim(np.datetime64("2025-09-26"), np.datetime64("2025-11-15"))
ax.set_xlabel("Date")
ax.set_ylabel("Number of events\nreceived at Rubin EFD")
ax.set_title(f"Total Events received during Rubin-LVK MDC: {max(cumulative)}")
fig.savefig("./CumulativeEventsDuringMDC.jpg", dpi=300)
```
